## 5. Data Filtering and Grouping

```python
df["City"].unique()
```

```python
df["City"].nunique()
```

```python
delhi_data = df[df["City"] == "Delhi"]

delhi_data.head()
```

```python
high_aqi = df[df["AQI"] > 300]

high_aqi.head()
```

```python
city_aqi = df.groupby("City")["AQI"].mean()

city_aqi
```

```python
city_aqi.sort_values(ascending=False)
```

```python
city_aqi.sort_values(ascending=False).head(10)
```

```python
city_aqi.sort_values().head(10)
```

## 6. Data Visualization

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df["AQI"].dropna(), bins=30)

plt.title("Distribution of AQI")
plt.xlabel("AQI")
plt.ylabel("Frequency")

plt.show()
```

```python
top10 = city_aqi.sort_values(ascending=False).head(10)
```

```python
plt.figure(figsize=(12, 6))

plt.bar(top10.index, top10.values)

plt.title("Top 10 Cities by Average AQI")
plt.xlabel("City")
plt.ylabel("Average AQI")

plt.xticks(rotation=45)

plt.show()
```

```python
clean10 = city_aqi.sort_values().head(10)

plt.figure(figsize=(12, 6))

plt.bar(clean10.index, clean10.values)

plt.title("Top 10 Cleanest Cities")
plt.xlabel("City")
plt.ylabel("Average AQI")

plt.xticks(rotation=45)

plt.show()
```
